```python
import json 
import datetime

import numpy as np 
import pandas as pd
import plotly.express as px
twitterData = pd.read_csv("data.csv", sep=',(?=\S)', engine='python')
twitterData.head()
```

```python
def delete_quotes(x):
    return x[1:-1]
for column in ["id", "screenName", "avatar", "lang", "tweetId"]:
    twitterData[column] = twitterData[column].apply(delete_quotes)
    
for column in ["tags", "friends"]:
    twitterData[column] = twitterData[column].apply(lambda x: json.loads(x))
twitterData.head()
```

```python
rank = 100
followersCount = -1

followersRank = []
followersCounts = []

while True:
    followersCount = twitterData[(twitterData["followersCount"] >= rank) & (twitterData["followersCount"] < rank * 10)].shape[0]
    if followersCount == 0:
        break
        
    followersRank.append(rank)
    followersCounts.append(followersCount)
    rank *= 10

followersCountByRankDict = {
    "count": followersCounts,
    "rank": followersRank
}
followersCountByRank = pd.DataFrame(data=followersCountByRankDict)

fig = px.scatter(followersCountByRank, x="rank", y="count")
fig.show()
```

```python
rank = 100
friendsCount = -1

friendRank = []
friendsCounts = []

while True:
    friendsCount = twitterData[(twitterData["friendsCount"] >= rank) & (twitterData["friendsCount"] < rank * 10)].shape[0]
    #print(friendsCount)
    if friendsCount == 0:
        break
        
    friendRank.append(rank)
    friendsCounts.append(friendsCount)
    rank *= 10

friendsCountByRankDict = {
    "count": friendsCounts,
    "rank": friendRank
}
#print(friendsCountByRankDict)
friendsCountByRank = pd.DataFrame(data=friendsCountByRankDict)

fig = px.scatter(friendsCountByRank, x="rank", y="count")
fig.show()
```

```python
fig = px.scatter(twitterData, x="followersCount", y="friendsCount", hover_data=["screenName"])
fig.show()
```

```python
def parse_date(x):
    return datetime.datetime.fromtimestamp(x/1000.0) \
            .strftime("%Y-%m-%d")


twitterData['lastSeenDatetime'] = twitterData['lastSeen'].apply(parse_date)
twitterData['lastSeenDatetime'].value_counts()
```

```python
fig = px.histogram(twitterData, x="lastSeenDatetime", nbins=100)
fig.show()
```

```python
fig = px.scatter(twitterData, x="lastSeenDatetime", y="followersCount")
fig.show()
```
